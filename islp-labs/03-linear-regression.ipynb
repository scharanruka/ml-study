### 3.6.2 Simple Linear Regression

```python
df = load_data("Boston")

df.head()
```

```python
X = pd.DataFrame({"intercept": np.ones(df.shape[0]), "lstat": df["lstat"]})

X
```

```python
y = df["medv"]
model = sm.OLS(y, X)
results = model.fit()
```
