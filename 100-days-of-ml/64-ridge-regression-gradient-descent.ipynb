```python
from sklearn.metrics import r2_score
from sklearn.datasets import load_diabetes
import numpy as np
```

```python
X, y = load_diabetes(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=4)
```

```python
from sklearn.linear_model import SGDRegressor

reg = SGDRegressor(
    penalty="l2", max_iter=500, eta0=0.1, learning_rate="constant", alpha=0.001
)
reg.fit(X_train, y_train)

y_pred = reg.predict(X_test)
r2_score(y_test, y_pred)
```

### Building own Ridge SGD

```python
class RidgeSGD:
    def __init__(
        self, alpha: float = 0.1, epochs: int = 100, learning_rate: float = 0.1
    ):
        self.alpha: float = alpha
        self.epochs: int = epochs
        self.lr: float = learning_rate

        self.coef_ = None
        self.intercept_ = None

    def fit(self, X_train, y_train):
        self.coef_ = np.ones(X_train.shape[1])
        self.intercept_ = 0
        theta = np.insert(self.coef_, 0, self.intercept_)

        X_train = np.insert(X_train, 0, 1, axis=1)

        for _ in range(self.epochs):
            theta_der = (
                np.dot(X_train.T, X_train).dot(theta)
                - np.dot(X_train.T, y_train)
                + self.alpha * theta
            )
            theta -= self.lr * theta_der

        self.intercept_ = theta[0]
        self.coef_ = theta[1:]

    def predict(self, X_test):
        return np.dot(X_test, self.coef_) + self.intercept_
```

```python
reg = RidgeSGD(epochs=500, alpha=0.001, learning_rate=0.005)

reg.fit(X_train, y_train)
y_pred = reg.predict(X_test)
r2_score(y_test, y_pred)
```
