```python
import numpy as np
from sklearn.datasets import load_diabetes
```

```python
X, y = load_diabetes(return_X_y=True)
X.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)

print(X_train.shape)
print(X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X_train, y_train)

y_pred = lr.predict(X_test)
```

```python
from sklearn.metrics import r2_score

r2_score(y_test, y_pred)
```

## Building own Linear Regression Class

```python
class MyLinearRegression:
    def __init__(self):
        self.coef_ = None
        self.intercept_ = None

    def fit(self, X_train, y_train):
        # adding the 1 column in the beginning
        # we are probably adding the 1 to multiply the intercept value in the B matrix
        X_train = np.insert(X_train, 0, 1, axis=1)

        # calculate the coeffs
        B = np.linalg.inv(np.dot(X_train.T, X_train)).dot(X_train.T).dot(y_train)

        self.intercept_ = B[0]
        self.coef_ = B[1:]

    def predict(self, X_test):
        y_pred = self.intercept_ + np.dot(X_test, self.coef_)
        return y_pred
```

```python
lr = MyLinearRegression()

lr.fit(X_train, y_train)
lr.intercept_
```

```python
y_pred = lr.predict(X_test)
r2_score(y_test, y_pred)
```
