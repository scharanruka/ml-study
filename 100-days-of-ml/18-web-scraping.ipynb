```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/107.0.0.0 Safari/537.36"
}

webpage = requests.get(
    "https://www.ambitionbox.com/list-of-companies?page=1", headers=headers
).text
```

```python
soup = BeautifulSoup(webpage, "lxml")
```

```python
companies_name_raw = soup.find_all("h2", class_="companyCardWrapper__companyName")
companies_n = []
for c in companies_name_raw:
    companies_n.append(c.text.strip("\n\t"))

companies_n
```

```python
reviews_raw = soup.find_all("div", class_="rating_text")
reviews_raw[0].text.strip()
```

```python
companies = soup.find_all("div", class_="companyCardWrapper")
for c in companies:
    name = c.find("h2").text.strip()
    rating = c.find("div", class_="rating_text").text.strip()
    review = (
        c.find("span", class_="companyCardWrapper__companyRatingCount")
        .text.strip()
        .strip("()")
    )
    print(f"{name} | {rating} | {review}")
```
