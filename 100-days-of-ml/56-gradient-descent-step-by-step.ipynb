```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
X, y = make_regression(
    n_samples=4, n_features=1, n_informative=1, n_targets=1, noise=80, random_state=13
)
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y)
```

```python
# Apply Ordinary Least Squares

from sklearn.linear_model import LinearRegression

lr = LinearRegression()
lr.fit(X, y)

print(lr.coef_, "|", lr.intercept_)
```

```python
plt.scatter(X, y)
plt.plot(X, lr.predict(X), color="red")
```

```python
# Let's apply Gradient Descent assuming slop is constant: m = 78.35
# b = 0
y_pred = ((78.35 * X) + 0).reshape(4)
```

```python
plt.scatter(X, y)
plt.plot(X, lr.predict(X), color="red", label="OLS")
plt.plot(X, y_pred, color="blue", label="b = 0")

plt.legend()
plt.show()
```

```python
m = 78.35
b = 0

loss_slope = -2 * np.sum(y - m * X.ravel() - b)
loss_slope
```

```python
# Let's take learning rate=0.1
learn_rate = 0.1

step_size = learn_rate * loss_slope
step_size
```

```python
# b_new = b_old - n*slope
b = b - step_size
b
```

```python
y_pred1 = ((m * X) + b).reshape(4)

plt.scatter(X, y)
plt.plot(X, lr.predict(X), color="red", label="OLS")
plt.plot(X, y_pred1, color="green", label=f"b = {b}")
plt.plot(X, y_pred, color="blue", label=f"b = {0}")

plt.legend()
plt.show()
```

---
### In loop form

```python
b = -100
m = 78.35
lr = 0.1

epochs = 10

for i in range(epochs):
    loss_slope = -2 * np.sum(y - m * X.ravel() - b)
    b = b - (lr * loss_slope)

    y_pred = m * X + b

    plt.plot(X, y_pred, label=f"b={b.round(2)}")

plt.legend()
plt.scatter(X, y)
```
