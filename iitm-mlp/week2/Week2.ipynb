### Lec-1 Introduction to `scikit-learn`

This lecture introduces Scikit-learn, a primary Python library for machine learning tasks. It streamlines the workflow through consistent APIs designed for data preprocessing, model training, and evaluation.

##### Core Design Principles
- Consistency: Uses a unified interface (e.g., fit, predict) across various estimators.
- Inspection: Models are designed to be inspectable, allowing users to understand parameters and internal states.
- Composability: Building blocks can be combined to form complex machine learning pipelines.

##### Primary Object Types
- Transformers: Responsible for data cleaning and feature transformation.
- Estimators: Core objects that learn parameters from data, conceptually minimizing a loss function.
- Predictors: Objects that utilize trained models to generate inferences or predictions.

##### Practical Application
The library provides modular support for classification, regression, clustering, and feature selection. Practitioners are encouraged to leverage the comprehensive documentation to understand specific module APIs and default parameter configurations for efficient model development.

### Lec-2 Data Loading

#### Overview of Scikit-Learn Data Loading
This lecture explores the various functionalities provided by Scikit-Learn to import datasets for machine learning workflows. Data loading is categorized into three primary approaches:

#### Core Loading Methods
1. Dataset Loaders: Ideal for smaller, built-in toy datasets that provide immediate access to features and labels.
2. Fetchers: Used to download and cache larger, real-world datasets from external internet sources.
3. Generators: Allow for the creation of synthetic datasets with specific properties for controlled experimentation.

##### Data Object Structure
Most loaders and fetchers return a dictionary-like object containing:
- Features (X): The input matrix of size 
- Targets (y): The label or response vector.

##### Handling External Formats
For custom data, Scikit-Learn integrates with tools to process various file formats:
- Text Files: Handled via utility functions like load_files.
- Standard Formats: Integration with libraries like NumPy and Pandas to import CSV, image, or video data, converting them into the numerical format required for model training.

### Lec-4 Data Pre-processing

Real-world data often requires cleaning and transformation before it can be used for model training. Scikit-learn provides a robust ecosystem for these tasks, ensuring consistency across training, testing, and evaluation sets through the use of Pipelines.

#### Key Preprocessing Steps
1. Feature Extraction: Converts raw data formats into numerical matrices. Tools include DictionaryVectorizer for categorical mapping and FeatureHasher for high-speed, memory-efficient vectorization.
2. Missing Value Imputation: Instead of discarding incomplete records, missing data is filled using:
    - SimpleImputer: Replaces gaps with statistics like mean, median, or the most frequent value.
    - KNNImputer: Estimates missing values by identifying the `k` nearest neighbors based on the Euclidean distance formula: 

#### Core Transformer Methodology
Every transformer follows a standardized interface:

- `fit`: Learns parameters (e.g., mean or standard deviation) from the training data.
- `transform`: Applies the learned parameters to scale or fill the dataset.
- `fit_transform`: A combined, more efficient execution of both methods.

### Lec-5 Categorical Transformers

This lecture explores methods for encoding categorical data into numeric formats suitable for machine learning algorithms using scikit-learn.

#### Encoding Techniques
1. OneHotEncoder: Converts categorical features into a binary array by creating one column for each unique value. If there are k  unique values, the resulting feature matrix expands to k columns.
2. LabelEncoder: Maps target labels to integers between 0 and `k-1`. It is designed for one-dimensional target vectors.
3. OrdinalEncoder: Similar to LabelEncoder but supports multi-dimensional feature matrices.

#### Specialized Binarizers
1. LabelBinarizer: Transforms multi-class target labels into a binary matrix format, useful for models requiring a one-versus-all setup.
2. MultiLabelBinarizer: Manages cases where a single sample contains multiple labels, mapping each unique class to a binary indicator column.
#### Data Augmentation

- Dummy Feature Addition: Often used to introduce a bias term into the dataset by appending a column vector containing exclusively ones.

### Lec-6 Numeric Transformers

Numerical preprocessing is essential for improving model performance and convergence during iterative optimization. This lecture covers three primary transformation categories:

1. Feature Scaling: Standardizing scales helps algorithms like gradient descent converge faster. Common techniques include:

    - **StandardScaler**: Centers data around a mean of 0 with unit variance using z = (x-mu)/sigma
    - **MinMaxScaler**: Scales features to a fixed range, typically [0,1] using x_scaled.
    - **MaxAbsScaler**: Scales data to the range [-1,1]  by dividing by the maximum absolute value.

2. Polynomial Transformers
These generate higher-order feature combinations, enabling models to capture non-linear relationships between inputs and targets.

3. Discretization
Techniques like `KBinsDiscretizer` convert continuous numerical variables into discrete bins. This is achieved by dividing the feature range into segments, which are then represented using ordinal or one-hot encoding, allowing models to interpret continuous values as distinct categories.

### Lec-7 Filter based Feature Selection

Feature selection improves machine learning models by removing irrelevant variables, thereby reducing dataset size and computational costs.

#### Core Approaches
1. Variance Threshold: Eliminates features with low variance, as they offer minimal predictive value.
2. Univariate Selection: Uses statistical tests to score features individually. Key methods include:
    - SelectKBest: Retains the top `k` features based on scoring functions.
    - SelectPercentile: Keeps a user-defined top percentage of high-scoring features.
    - GenericUnivariateSelect: A flexible approach allowing various selection strategies.

#### Scoring Functions
Univariate selection requires a metric to evaluate feature importance:
- Mutual Information: Measures dependency between variables using information theory principles. If `I(X;Y)=0`, the variables X
 and Y are independent.
- Chi-Square:  Assesses dependence between non-negative features and class labels, ideal for classification tasks with sparse data.

#### Important Considerations
Always ensure the scoring function aligns with the problem type. For instance, do not apply regression-based scoring to classification problems to avoid producing invalid results.

### Lec-8 Wrapper based Feature Selection

Wrapper methods evaluate subsets of features by using a machine learning estimator, unlike filter methods that rely on independent scoring functions. They are iterative and often more computationally intensive.

#### Key Techniques
1. Recursive Feature Elimination (RFE): Recursively fits an estimator on all features, removes the least important feature based on its weight or importance attribute, and repeats until the target number of features is reached.
2. RFECV: An extension of RFE that employs cross-validation to automatically determine the optimal number of features.
3. SelectFromModel: Trains an estimator once and selects features based on an importance threshold, such as non-zero coefficients in 
$L_1$-regularized models.

#### Sequential Feature Selection
This approach uses greedy search strategies to add or remove features one at a time:

1. Forward Selection: Starts with zero features and iteratively adds the feature that yields the best cross-validation score.
2. Backward Selection: Starts with all $M$ features and iteratively removes the least important ones until the desired count is achieved.

Efficiency depends on the number of features. Backward selection is often more efficient when the target number of features is close to the total set size $M$

### Lec-9 Heterogenous Feature Transformations

Data preprocessing often requires applying specific transformations to different subsets of features. Scikit-learn provides composite transformers to handle these diverse requirements efficiently.

#### ColumnTransformer
This tool enables the application of distinct preprocessing techniques to different feature columns, such as:

- Numerical features: Applying scaling methods like MaxAbsScaler.
- Categorical features: Using OneHotEncoder to convert labels into numerical representations.

#### TransformedTargetRegressor
This class handles transformations on the target variable ($y$) before model training, which is useful when the target distribution requires normalization or log-scaling.

1. Transformation: The target $y$ is mapped to a transformed space using a function, $f(y)$.
2. Regression: A regressor is fitted on features $X$ and the transformed target $f(y)$.
3. Inverse Transformation: Predictions made in the transformed space are mapped back to the original scale using an inverse function. 

By integrating these tools into a pipeline, one can create robust machine learning workflows that maintain interpretability while handling heterogeneous input data.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import MaxAbsScaler, OneHotEncoder


column_transformer = ColumnTransformer(
    [
        ('ageScaler', MaxAbsScaler(), [0]),
        ('genderEncoder', OneHotEncoder(dtype='int'), [1])

    ], remainder='drop',verbose_feature_names_out=False
)
column_transformer
```

### Lec-10 Dimensionality Reduction by PCA

PCA is a fundamental dimensionality reduction technique used in machine learning to simplify high-dimensional datasets while retaining the most significant information.

#### Core Methodology
The technique transforms data into a new coordinate system by identifying directions, known as principal components, that capture the maximum variance. The process follows these steps:

1. Data Projection: Points are projected onto vectors to minimize information loss.
2. Variance Maximization: The first principal component (PC1) is selected to account for the highest variance. Subsequent components (PC2,PC3...) are chosen to capture remaining, orthogonal variance.
3. Decomposition: The process leverages Singular Value Decomposition (SVD) to mathematically decompose the feature matrix.

#### Application
PCA is particularly effective in unsupervised learning scenarios where the goal is to reduce the number of features without losing the essential structure of the data. By focusing on these components, practitioners can optimize model performance and visualize complex, multi-featured datasets in a reduced-dimensional space.
