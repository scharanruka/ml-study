### Lec-1 Classification Functions in scikit learn

This lecture explores the scikit-learn (sklearn) API for classification, focusing on both specific and generic algorithmic implementations. The primary goal is to provide practical knowledge for training, predicting, and evaluating machine learning models.

#### Classification APIs
Sklearn provides two main types of classification APIs:

- Specific APIs: Designed for individual algorithms like Ridge Classifier, Perceptron, and Logistic Regression. These use specialized solvers for optimization.
- Generic APIs: The SGDClassifier is a versatile tool that uses Stochastic Gradient Descent. By adjusting the loss function parameter, it can mimic various specific algorithms.

#### Common Methods
All classification estimators in sklearn share a standard set of methods:

- `fit(X, y)`: Trains the model using feature matrix X and label vector y.
- `predict(X)`: Generates class labels for new data.
- `decision_function(X)`: Returns confidence scores.
- `score(X, y)`: Evaluates model accuracy.

##### Key Algorithms
1. **Ridge Classifier**: Useful for least squares classification. It converts binary targets to -1 and 1, then performs regression. It supports various solvers like SVD, LSQr, and SAG.
2. **Perceptron**: A simple, efficient algorithm for large-scale learning. It is essentially an SGDClassifier with specific loss and learning rate settings.
3. **Logistic Regression**: Optimizes a regularized cross-entropy loss. It supports multiple solvers (e.g., liblinear, lbfgs, saga) and allows for L1, L2, or Elastic Net regularization. The parameter C acts as the inverse of the regularization rate, where smaller values increase regularization strength.

#### Advanced Customization
- Regularization: Algorithms often support L1, L2, or Elastic Net penalties to improve numerical stability and prevent overfitting.
- Handling Imbalance: The class_weight parameter allows users to assign different penalties to classes, helping manage imbalanced datasets.
- Optimization Solvers: The choice of solver is critical and depends on data size, scaling, and whether the task is binary or multi-class. SGD-based solvers (like SAG or SAGA) perform best when features are standardized.

### Lec-2 Evaluating Classifiers

This lecture covers techniques for assessing classification models in scikit-learn.

#### Model Evaluation Methods
- Stratified Cross-Validation: Uses methods like StratifiedKFold to maintain class distribution across folds, addressing class imbalance.
- Hyperparameter Tuning: LogisticRegressionCV enables automated optimization of regularization parameters.

#### Key Metrics
1. Core Metrics: Accuracy, Precision, Recall, and F1-Score.
2. Advanced Tools: Confusion Matrix, ROC-AUC, and Precision-Recall curves visualize performance.

#### Multi-class/Multi-label Extension
Metrics are extended via averaging methods:

- Macro: Arithmetic mean.
- Weighted: Class-size weighted average.
- Micro: Sample-class pair equality.

### Lec-3 Demo: Binary Class Image Classification using Perceptron

This lecture focuses on applying the Perceptron algorithm to binary image classification using the MNIST dataset. The primary goal is to build a detector that identifies the handwritten digit '0'.

#### Data Preparation
To prepare the data for the Perceptron, the following steps are performed:

1. Data Loading: Retrieving 70,000 samples of 28x28 grayscale images from MNIST.
2. Normalization: Scaling pixel values from the range 0-255 to 0-1 using MinMaxScalar.
3. Label Modification: Transforming the multi-class dataset into a binary one, where the digit '0' is assigned a label of +1 and all other digits are assigned -1.

#### Model Training and Evaluation
The classification process follows these stages:

1. Baseline: Establishing a performance benchmark using a Dummy Classifier with a 'most frequent' strategy.
2. Training: Implementing the Perceptron model, which uses a linear combination of weights and features passed through a sign-based activation function.
3. Evaluation: Assessing the model using metrics like Precision, Recall, and the F1-Score. Confusion matrices and ROC curves are used to analyze performance and the trade-off between precision and recall.
4. Optimization: Exploring Cross-Validation to ensure model robustness and the Warm Start parameter to efficiently continue training from previous iterations.

#### Key Mathematical Concept
The Perceptron loss function for an example is defined as:
`Max(0, -w^T * x * y)`

This guides the weight update rule to ensure the algorithm converges if the data is linearly separable.

### Lec-5 Demo: MNIST Digits using SGDRegressor

This lecture demonstrates how to implement a binary digit classification task on the MNIST dataset using Logistic Regression via the SGDClassifier. The objective is to identify if a handwritten image represents the digit 0.

#### Data Preprocessing
- Feature Scaling: Pixel intensities (0-255) are normalized to the range [0, 1] using MinMaxScaler, which is essential for stable gradient-based optimization.
- Binary Setup: The dataset is modified to create a binary label: 1 for images of the digit 0 and 0 for all other digits.

#### Model Implementation
1. Baseline Model: A DummyClassifier is used as a benchmark, achieving 90% accuracy by predicting the most frequent class (non-zero digits).
2. Logistic Regression: The SGDClassifier is configured with loss='log' to perform logistic regression. The model computes a linear combination of features (z = w1x1 + ... + wnxn + b) and passes it through the sigmoid function (1 / (1 + e^-z)).
3. Optimization: The model minimizes cross-entropy (log loss) using stochastic gradient descent. The lecture illustrates the importance of learning rates and the use of warm starts to visualize learning curves.

#### Evaluation and Hyperparameter Tuning
- Metrics: Beyond accuracy, performance is assessed using precision, recall, F1-score, and confusion matrices to handle potential class imbalances.
- Cross-Validation: RandomizedSearchCV is utilized to tune the learning rate, ensuring the model generalizes well across different subsets of the data. Evaluation curves, including Precision-Recall and ROC curves, confirm high classification performance.
