### Lec-1 Large Scale Machine Learning

This lecture addresses the challenge of training machine learning models on datasets that exceed available RAM. Traditional methods that load entire datasets into memory are impractical for massive real-world data, requiring strategies for incremental or out-of-core learning.

#### Incremental Learning Strategies

Scikit-learn facilitates large-scale learning primarily through the partial_fit method. Unlike the standard fit method, which requires the full dataset, partial_fit allows for iterative updates using data batches. Key points include:

*   **Batch Processing:** Data is loaded and processed in smaller chunks, updating model parameters incrementally until the entire dataset is covered.
*   **Use Cases:** This approach is essential for massive datasets that do not fit in memory or for streaming data where new batches arrive over time.
*   **Compatible Estimators:** Algorithms like SGDClassifier, SGDRegressor, and MiniBatchKMeans support this incremental training workflow.

#### Incremental Preprocessing

Processing text at scale requires specialized tools to avoid memory bottlenecks. The lecture compares two main vectorizers:

1.  **CountVectorizer:** Useful for smaller datasets but requires storing a complete vocabulary in memory.
2.  **HashingVectorizer:** Ideal for large-scale tasks. It maps tokens directly to columns using a hashing function, eliminating the need to store a vocabulary and reducing memory overhead.

#### Practical Workflow

The session demonstrates a sentiment analysis task combining HashingVectorizer and SGDClassifier. By processing text in batches, the model learns incrementally, providing a scalable pipeline for real-time applications where data memory constraints are a critical factor.

### Lec-2 Confusion Matrix

A **confusion matrix** is a fundamental tool for evaluating classification models by comparing predicted labels against actual outcomes.



#### Core Components

1.  **Binary Classification**: Analyzes four outcomes:
    *   **True Positives (TP)** and **True Negatives (TN)** represent correct predictions.
    *   **False Positives (FP)** and **False Negatives (FN)** indicate misclassifications.

#### Model Evaluation

*   **Multi-class Extension**: The matrix expands to an N x N grid, tracking instances across multiple categories.
*   **Insight**: It reveals specific systematic errors, biases, or class imbalances that simple accuracy metrics often overlook, helping guide necessary model improvements.

### Lec-3 Accuracy

#### Definition
Accuracy measures the fraction of correct predictions made by a model. It is calculated as:

*   **Accuracy** = (Number of Correct Predictions) / (Total Number of Predictions)

#### Key Concepts

*   **Binary Classification:** Accuracy is defined as (True Positives + True Negatives) / (Total Predictions).
*   **Imbalanced Data:** Accuracy can be misleading. A model predicting the majority class for everything may achieve high accuracy despite failing to identify minority classes.
*   **Evaluation:** In practice, complement accuracy with a confusion matrix and other metrics to ensure comprehensive model performance analysis.

### Lec-4 Precision

Precision is a vital evaluation metric for classification models, particularly when minimizing false positives is critical. It calculates the accuracy of positive predictions.

#### Key Concepts

*   **Definition:** Precision is the ratio of true positives to the sum of true positives and false positives.
*   **Formula:** Precision = True Positives / (True Positives + False Positives)

#### Real-World Applications

1.  **Medical Diagnosis:** Avoiding unnecessary treatments from false alarms.
2.  **Fraud Detection:** Ensuring customer transactions are not incorrectly blocked.

Techniques demonstrated include using the scikit-learn library and manual confusion matrix derivation.

### Lec-5 F1 Score

- The F1 score is a crucial metric for evaluating classification models, especially when dealing with imbalanced datasets. It provides a balanced view by combining precision and recall into a single value.

#### Core Formula

The F1 score is the harmonic mean of precision and recall:

` F1 = 2 * (Precision * Recall) / (Precision + Recall) `

#### Key Takeaways

1.  It is calculated using true positives, false positives, and false negatives from a confusion matrix.
2.  Implementation is straightforward in Python using the scikit-learn library by passing true and predicted labels.

### Lec-6 Recall

Recall is a key classification metric, also known as sensitivity or the true positive rate. It measures a model's ability to identify all positive instances correctly.

#### Formula

`Recall = True Positives / (True Positives + False Negatives)`

#### Application

1.  **Use Cases**: Critical for scenarios like fraud detection or disease diagnosis where missing a positive case (a false negative) is costly.
2.  **Implementation**: In Python, the recall_score function from scikit-learn computes this metric by comparing ground truth values against model predictions.

By minimizing false negatives, recall ensures high-stakes positive outcomes are prioritized.

### Lec-7 Multiclass

This lecture explains how to extend binary classification metrics to problems with more than two classes.

#### Key Evaluation Metrics

*   **Accuracy**: The ratio of correct predictions to total instances, calculated using the confusion matrix diagonal.
*   **Precision**: The ratio of true positives to the sum of true positives and false positives for each class.
*   **Recall**: The ratio of true positives to the sum of true positives and false negatives for each class.
*   **F1-Score**: The harmonic mean of precision and recall.

#### Reporting Results

1.  Use the **classification report** for a summary of per-class metrics.
2.  **Macro average**: Calculates the unweighted mean across classes.
3.  **Weighted average**: Calculates the mean weighted by class support.
