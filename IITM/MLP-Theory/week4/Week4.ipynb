### Lec-1 Linear Regression

#### Baseline and Core Models

### Lec-2 Model Evaluation

Effective model evaluation is critical for understanding performance and preventing overfitting or underfitting. The process involves splitting data into training and test sets to assess empirical and generalization errors.

#### Key Evaluation Metrics
Scikit-learn provides several metrics to assess performance:

- R-squared (Coefficient of Determination): Calculated as 1 - (u / v), where u is the residual sum of squares and v is the total sum of squares. A score of 1 indicates a perfect fit, 0 represents a constant model, and negative values imply the model is worse than a baseline.
- Error Metrics: Common metrics include:
    1. Mean Absolute Error (MAE): Average absolute difference between predicted and actual values.
    2. Mean Squared Error (MSE): Average of squared errors.
    3. Max Error: The worst-case error for a single prediction.

#### Robust Model Assessment
To ensure results are not dependent on a single data split, cross-validation techniques are employed:

- K-Fold: Divides data into K subsets for rotation-based training and testing.
- Leave-One-Out: A special case of K-Fold where each sample serves as a test set.
- ShuffleSplit: Performs random permutations of data to generate multiple train-test splits.

#### Model Diagnostics
To diagnose performance issues, users can:

1. Use `cross_validate` to obtain detailed metrics and track fit times or estimator objects.
2. Utilize learning curves to visualize how the size of the training set affects error rates, which helps identify if more data or hyperparameter tuning is required.

### Lec-3 Polynomial Regression

Polynomial regression models non-linear relationships using a two-step pipeline process:

1. Feature Transformation: Apply polynomial features to the input matrix.
2. Linear Estimation: Fit a model (e.g., Linear Regression or SGDRegressor) on the transformed data.

#### Customization
- Interaction Features: Use the interaction_only=True parameter to include only product terms (like x1 * x2) while excluding individual higher-degree terms (like x1^2).
- Hyperparameter Tuning: Determining the optimal polynomial degree requires systematic tuning to balance model complexity and performance.

```python
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import PolynomialFeatures

poly_model = Pipeline([
    ('poly_transform', PolynomialFeatures(degree=2)),
    ('linear_regression', LinearRegression())
])
poly_model
```

### Lec-4 Regularization

This lecture covers techniques to prevent overfitting in regression models using scikit-learn.

#### Key Methods
- Ridge Regression (L2): Adds a penalty equal to the square of the magnitude of coefficients.
- Lasso Regression (L1): Adds a penalty equal to the absolute value of coefficients.

#### Implementation Strategies
1. Use specialized estimators like Ridge, Lasso, or SGDRegressor with specific penalties.
2. Tune the regularization strength (alpha) using cross-validation (GridSearchCV or RandomizedSearchCV).
3. Integrate into pipelines for polynomial regression.

### Lec-5 Hyper Parameter Tuning

Hyperparameter tuning is essential for optimizing machine learning model performance by selecting the best set of configuration variables that are not learned directly from the data.

#### Core Approaches
There are two primary automated methods for searching the hyperparameter space:

1. Grid Search: Exhaustively evaluates a defined grid of specific parameter values to identify the optimal configuration.
2. Randomized Search: Samples parameters from a specified distribution, which is often more efficient than grid search when dealing with large or continuous search spaces.

#### Model Evaluation and Optimization
Effective tuning requires a rigorous evaluation framework:

1. Data Splitting: Datasets are partitioned into training, validation, and test sets to prevent overfitting and ensure reliable model selection.
2. Cross-Validation: This technique validates model performance across multiple subsets of data, ensuring the chosen hyperparameters generalize well.

#### Practical Application
The process often involves model-specific tuning for algorithms like Lasso, Ridge, or ElasticNet. For instance, when determining the optimal degree for Polynomial Regression, the model is tested against various degrees to balance complexity and predictive accuracy, ultimately improving the model's overall generalization capability.

### Lec-6 Multi Learning Classification

This lecture explores how scikit-learn extends standard classification to handle complex scenarios where output labels exceed simple binary or single-class formats. These problems are categorized into three setups:

1. Multi-class Classification: Exactly one label per sample from more than two categories.
2. Multi-label Classification: Multiple labels per sample, with two total possible labels.
3. Multi-output Classification: Multiple labels per sample, where the total number of labels exceeds two.

#### Meta-Estimator Strategies
To manage these tasks, scikit-learn utilizes meta-estimators that decompose complex problems into simpler binary tasks:

- One-vs-Rest (OvR): Fits one binary classifier per class to distinguish it from all others.
- One-vs-One (OvO): Fits one classifier for every possible pair of classes.
- Classifier Chains: Arranges binary classifiers in a sequence to capture correlations between targets, where input for subsequent models includes outputs from previous ones.

#### Data Handling
Tools like `LabelBinarizer` are used to convert labels into indicator matrices of shape (n, k), where n is the number of samples and k is the number of classes, facilitating consistent processing across different multi-learning architectures.

```python
import numpy as np
from sklearn.preprocessing import LabelBinarizer
y = np.array(['bear', 'wolf', 'crow', 'deer'])
y_dense = LabelBinarizer().fit_transform(y)
y_dense
```

### Lec-8 K-Nearest Classifier

KNN is a non-parametric, instance-based learning method for classification that relies on storing training data rather than building an explicit model.

#### Core Classifiers
1. KNeighborsClassifier: Predicts labels based on the K-nearest neighbors; the default is K=5.
2. RadiusNeighborsClassifier: Predicts labels based on neighbors within a fixed radius (r), useful for non-uniformly sampled data.

#### Key Parameters
- Weights: Determines influence. Options include 'uniform' (equal weight) or 'distance' (closer points have more influence).
- Algorithms: Efficient search methods include Ball Tree, KD Tree, or Brute Force.
- Distance Metrics: Common choices include Euclidean, Manhattan, or Minkowski.

### Lec-9 Support Vector Machines (SVM) in scikit-learn

This lecture covers SVM implementation in scikit-learn for classification.

#### Key Estimators
- SVC and NuSVC: General-purpose classifiers based on LIBSVM.
- LinearSVC: A faster implementation using LIBLINEAR, optimized for linear kernels.

#### Core Concepts
- Regularization: Controlled by parameter C; strength is inversely proportional to its value.
- Kernels: Methods include linear, polynomial, RBF, and sigmoid. Gamma defines kernel coefficients.

SVMs are memory-efficient, performing well in high-dimensional spaces by utilizing a subset of training points called support vectors.

### Lec-10 Decision Trees

Decision trees are non-parametric supervised learning models used for classification and regression. Key points include:

#### Core Concepts
- Algorithms: Common types include ID3, C4.5, C5.0, and CART.
- Scikit-Learn: Utilizes an optimized CART implementation.

#### Key Parameters
- Splitter: Determines node split strategy (best or random).
- Criteria: Measures split quality; uses Gini or entropy for classification and squared_error for regression.
- Pruning: Prevents overfitting via pre-pruning or post-pruning techniques.

#### Practical Tips
- Perform dimensionality reduction if features are large.
- Balance datasets to avoid bias.
- Use `min_samples_split` or `min_samples_leaf` to control tree growth.

### Lec-11 Voting, Bagging & Random Forest

This lecture covers ensemble techniques in scikit-learn to improve model performance and reduce overfitting.

#### Core Topics
- **Voting Estimators**: Combines classifiers or regressors using hard or soft voting.
- **Bagging**: Enhances stability by training multiple estimators on bootstrap samples drawn with replacement.
- **Random Forest**: Builds a forest of decision trees by bagging and selecting feature subsets.

#### Key Parameters
- `n_estimators`: Defines the total number of trees.
- `max_features`: Determines feature selection for splits (e.g., sqrt, log2).
- `criterion`: Specifies splitting quality (Gini for classification, squared error for regression).
