### Lec-1 Chaining Transformers

This lecture explores how to build structured, reproducible machine learning workflows in scikit-learn using two primary constructs:

#### Pipeline API
- The Pipeline class chains multiple transformers and a final estimator into a single object.
- This approach simplifies the process by automating data flow and reducing code complexity.

- Integration: It allows for joint hyperparameter tuning using tools like GridSearchCV.
- Parameter Access: Individual steps are accessible via dictionary-like syntax or names, and parameters are tuned using the syntax `estimator_name__parameter_name`.
- Computation: The memory parameter enables caching of expensive transformations to improve efficiency during cross-validation.

#### Feature Union
FeatureUnion extends the workflow by allowing parallel data processing. It applies multiple transformers to the input data simultaneously and concatenates their outputs into a single, wider feature matrix. This is particularly useful for applying different preprocessing techniques (like LabelBinarizer vs. StandardScaler) to categorical and numerical features before merging them into a unified feature set for modeling.

### Lec-2 Demonstration of data extraction, imputation, scaling, visualizing feature distribution

This lecture demonstrates essential data preprocessing techniques to prepare raw datasets for effective machine learning model training.

#### Core Preprocessing Stages
1. Feature Extraction: Transforms unstructured data, such as dictionaries, into structured matrices using tools like DictionaryVectorizer.
2. Data Cleaning: Addresses missing values via SimpleImputer using strategies like mean, median, or constant filling to ensure full feature matrices.
3. Feature Scaling: Adjusts numerical features to a unified scale, which is critical for algorithms sensitive to magnitude, such as Gradient Descent or SVM.


#### Scaling Methodologies
Normalization and standardization are two primary approaches used to manipulate feature distributions:

- Normalization: Constrains data to a specific range
- Standardization: Transforms data to have a mean of $0$ and a standard deviation of 1 using the StandardScaler

These steps ensure that diverse features contribute appropriately to model convergence and distance-based calculations

### Lec-3 Demonstration of Data transformation, Composite Transformers

This lecture explores essential preprocessing techniques in scikit-learn for preparing datasets for machine learning workflows. Effective feature engineering is critical for improving model performance and interpretability.

#### Core Transformation Techniques
- Dummy Features: The add_dummy_feature utility augments datasets by injecting a bias column, which consists of constant values (e.g.,1) for every sample.
- Custom Transformations: FunctionTransformer allows the integration of arbitrary Python functions, such as $y = log(1+x)$ to handle heterogeneous data types or non-linear scaling.
- Polynomial Features: Used to expand input spaces by generating interaction and power terms, such as $a^2$ for an input sample.


#### Discretization and Encoding
1. Discretization: KBinsDiscretizer partitions continuous features into discrete bins, often combined with one-hot encoding to enhance model expressiveness.
2. Categorical Encoding: Various methods are employed to convert nominal categories into numerical formats:
    - OneHotEncoder: Creates binary vectors for unique categories.
    - OrdinalEncoder: Assigns unique integer labels.
    - MultiLabelBinarizer: Handles multiple labels per sample.
3. Composite and Target Transformations
    - ColumnTransformer: Enables the application of heterogeneous transformations on different feature subsets simultaneously.
    - TransformedTargetRegressor: Facilitates the scaling or transformation of target variables before regression, with built-in mechanisms for inverse transformations.

### Lec-4 Demonstration of feature selection demonstration, PCA, pipelines, handling class imbalance

This lecture covers essential scikit-learn utilities for streamlining machine learning workflows, focusing on feature selection, dimensionality reduction, pipeline management, and addressing class imbalance.

#### Feature Selection
Techniques are categorized into two main groups:

1. Filter-based methods: Use statistical measures to select features. Examples include VarianceThreshold, SelectKBest, SelectPercentile, and GenericUnivariateSelect.
2. Wrapper-based methods: Use an estimator to rank features. These include Recursive Feature Elimination (RFE), SelectFromModel, and Sequential Feature Selection.

#### Dimensionality Reduction
Principal Component Analysis (PCA) is used to reduce the feature space while maximizing retained variance. It transforms input data into orthogonal components, where the first component captures the largest variance

#### Pipelines and FeatureUnions
To ensure consistent data transformation across training and testing sets, scikit-learn provides:

1. Pipelines: Chain sequences of transformers and a final estimator. The final step must implement a fit method, while intermediate steps must implement fit and transform.
2. FeatureUnions: Apply multiple transformers in parallel, concatenating their outputs.

#### Handling Class Imbalance
Imbalanced datasets lead to biased model performance. Strategies to mitigate this include:

- Undersampling: Reducing the majority class samples (e.g., RandomUnderSampler).
- Oversampling: Increasing minority class samples (e.g., RandomOverSampler or SMOTE—Synthetic Minority Over-sampling Technique) to create a more balanced distribution.
