```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("./DailyDelhiClimateTest.csv")

df = df[["date", "meantemp"]]
print(df.shape)
df.head()
```

```python
plt.scatter(df["date"], df["meantemp"], color="red")
```

```python
alpha = 0.1
```

```python
x1 = df["meantemp"].ewm(alpha=alpha).mean()
df["ewma"] = x1
df.head()
```

```python
plt.scatter(df["date"], df["meantemp"], color="red")
plt.plot(df["date"], x1, color="black", label=f"alpha = {alpha}")
plt.legend()
plt.show()
```
