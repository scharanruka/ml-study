This video covers fundamental **Natural Language Processing (NLP)** techniques used to convert human language into numerical formats suitable for machine learning models, specifically for tasks like **sentiment analysis**.

#### Core Text Representation Techniques

1. **One Hot Encoding**
    
    - Represents each word as a unique vector of zeros with a single one.
    - **Limitation:** Fails to capture the **semantic relationship** between words and results in sparse, high-dimensional matrices.
2. **Bag of Words (BoW)**
    
    - Creates a vocabulary of all unique words and represents documents by counting the frequency of these words.
    - **Limitation:** Ignores word order and **semantic meaning**.
3. **N-grams (Uni-grams, Bi-grams, Tri-grams)**
    
    - Extends BoW by taking sequences of _N_ words together (e.g., pairs of words for bi-grams).
    - **Advantage:** Captures context and word order better than BoW.
    - **Disadvantage:** Significantly increases the **dimensionality** of the data and computational complexity.
4. **Tf-Idf (Term Frequency-Inverse Document Frequency)**
    
    - Assigns weights to words based on their frequency within a specific document relative to their rarity across the entire corpus.
    - **Key Idea:** Words frequent in a document but rare overall are given higher importance.
    - **Advantage:** Better identifies significant words compared to simple frequency counts.

#### Custom Feature Engineering

Beyond automated techniques, practitioners often create **hand-crafted features** based on domain knowledge to improve model performance. Examples include:

- Counting specific **positive or negative words**.
- Calculating the **ratio** of positive to negative words.
- Measuring document **length** (word count or character count).

#### Conclusion

While traditional techniques like BoW and Tf-Idf are foundational, they often struggle with capturing deep semantic relationships, a limitation addressed by more advanced techniques like _Word2Vec_.
